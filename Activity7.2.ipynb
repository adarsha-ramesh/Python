```python
import pandas as pd

data = {
    "ProductID": [101, 102, 103, 104, 105, 106],
    "ProductName": [
        "Laptop", 
        "Smartphone", 
        "Desk Chair", 
        "Notebook", 
        "Headphones", 
        "Water Bottle"
    ],
    "Category": [
        "Electronics", 
        "Electronics", 
        "Furniture", 
        "Stationery", 
        "Electronics", 
        "Accessories"
    ],
    "Price": [75000, 30000, 8500, 50, 2500, 300],
    "Stock": [15, 25, 40, 200, 60, 150]
}

df = pd.DataFrame(data)
df
```

Filter the DataFrame to show only products in the 'Electronics' category.
Sort the filtered DataFrame by price in descending order.

```python
new=df[df["Category"]=="Electronics"]
new
```

```python
print(new.sort_values("Price", ascending=False))
```

groupby() on the category column and then use the sum() method on the sales
The final output should show the total sales for each category.

```python
print(df.groupby("Category")["Price"].sum())
```

Merge the two DataFrames on the productID column to create a single table that shows each order with the customer's name.

```python
data_customers = {
    "ProductID": [101, 102, 103, 104, 105, 106],
    "CustomerName": [
        "Amelia Johnson",
        "Oliver Smith",
        "Sophia Wilson",
        "Jack Taylor",
        "Isla Brown",
        "Charlie Lewis"
    ]
}

df_customers = pd.DataFrame(data_customers)

print(df_customers)
```

```python
import pandas as pd
new_df =pd.merge(df, df_customers, on="ProductID")
a=new_df.groupby("CustomerName")["ProductName"].apply(list)
a
```
