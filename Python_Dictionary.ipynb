A list of dictionaries are given, where each dictionary represents a customer with information like name, age, and purchase_amount. 
Write a single Python script that 
iterates through the list, 
performs a series of calculations (e.g., finds the average purchase amount), 
and prints the results.

```python
customers = [
    {
        "name": "Alice",
        "age": 28,
        "purchase_amount": 250.75
    },
    {
        "name": "Bob",
        "age": 35,
        "purchase_amount": 120.50
    },
    {
        "name": "Charlie",
        "age": 22,
        "purchase_amount": 89.99
    },
    {
        "name": "Diana",
        "age": 40,
        "purchase_amount": 560.00
    }
]
```

```python
import pandas as pd
df = pd.DataFrame(customers)
df
```

```python
round(df["purchase_amount"].mean())
```

```python
df["purchase_amount"].median()
```

```python
total1= 0
total2=0
for c in customers:
    total1+=c["purchase_amount"]
for c in customers:
    total2 +=1

print(total1/total2)
```

```python
#Find the highest purchase amount.
amount=[]

for c in customers:
    amount.append(c["purchase_amount"])
    amount.sort(reverse=True)

print("highest purchase amount is", amount[0])
```

```python
#Find the customer with highest purchase amount.
max_customer = customers[0]

for c in customers:
    if c["purchase_amount"] > max_customer["purchase_amount"]:
        max_customer = c

print("Customer with highest purchase:", max_customer)
```

```python
#Count how many customers are older than 30.

older=[]

for c in customers:
    if c["age"]>=30:
        older.append(c)

print("There are ",len(older), "customers older than 30")
```

```python
#Print names of customers who spent more than 200

customer_name =[]

for c in customers:
    if c["purchase_amount"] >200:
        customer_name.append(c["name"])

print(customer_name)
```

```python
#Sort customers by purchase_amount (descending)
new_dict={}

for c in customers:
    new_dict.update({c["name"]:c["purchase_amount"]})

print(sorted(new_dict.items(), key=lambda item: item[1], reverse = True))    
    
```
