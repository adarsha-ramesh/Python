```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

file= pd.read_csv("Topic6assignment2 (1).csv")
file.isnull().sum()
```

```python
file.duplicated().sum()
```

```python
#Daily trend of Revenue over time
date= file["Date"]
revenue = file["Revenue"]
plt.figure(figsize=(15,6))
plt.plot(date, revenue, color= "blue")
plt.xlabel("Time")
plt.ylabel("Revenue")
plt.title("Daily trend of Revenue over time")
```

```python
#Compare Revenue between two specific products (Mobile vs Laptop)

item1 = file[file["Product"] == "Mobile"]
item2 = file[file["Product"] == "Laptop"]

mobile_revenue = item1["Revenue"].sum()
laptop_revenue = item2["Revenue"].sum()

products = ["Mobile", "Laptop"]
revenues = [mobile_revenue, laptop_revenue]

plt.bar(products, revenues, color="orange")  #plt.bar() needs numeric values
plt.xlabel("Products")
plt.ylabel("Revenue")
plt.title("Revenue between Mobile vs Laptop")
```

```python
#What percentage of total Units Sold is contributed by each Product?

file1 = file[file["Product"]=="Mobile"]
file2 = file[file["Product"]=="Laptop"]
file3 = file[file["Product"]=="Tablet"]

a= file1["Units_Sold"].sum() #units_by_product = file.groupby("Product")["Units_Sold"].sum()
b= file2["Units_Sold"].sum()
c= file3["Units_Sold"].sum()

unitssold = [a,b,c]
y = ["Mobile","Laptop", "Tablet"]

plt.pie(unitssold, labels=y, autopct="%1.1f%%")
plt.title("percentage of total Units Sold is contributed by each Product")
```

```python
#correlation between higher Customer Rating and Units_Sold

rating= file["Customer_Rating"]
units= file["Units_Sold"]

plt.scatter(rating, units, color="red")
plt.gca().set(xlabel="Ratings", ylabel="Units_Sold", title ="Correlation")
```
