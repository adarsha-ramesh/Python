```python
#BarChart
import matplotlib.pyplot as plt
import seaborn as sns

items = ["pen", "pencil", "Scale", "Eraser"]
quantity = [3,5,1,7]

fig, axes = plt.subplots(1, 2, figsize=(10,4))

axes[0].bar(items, quantity, color = "pink")
axes[0].set_title("Bar Chart using Matplotlib")

sns.barplot(x=items, y=quantity, ax= axes[1])
plt.title("Bar Chart using Seaborn")
```

```python
#LineChart
import matplotlib.pyplot as plt
import seaborn as sns
x = [1,2,3,4,5]
y = [2,4,6,8,10]

fig, axes = plt.subplots(1, 2, figsize=(10,4))

axes[0].plot(x, y, color = "pink")
axes[0].set_title("Line Plot using matplot")

sns.lineplot(x=x, y=y, ax= axes[1])
plt.title("Line Chart using Seaborn")
```

```python
#PieChart
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = ["A", "B", "C", "D", "E"]

plt.pie(x, labels=y)
plt.title("Pie chart using matplot")

#Seaborn dosn't support Pie chart
```

```python
#PieChart
import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y = [20,30,40,50,60]

plt.scatter(x,y, color = "pink")
plt.title("ScatterPlot using matplot")
```

```python
#Histogram
import numpy as np
data = np.random.randn(200)

plt.hist(data, bins=10, color = "pink")
plt.title("Histogram using matplotlib")
plt.show()
```

```python
#Histogram
import numpy as np
import seaborn as sns
data = np.random.randn(200)

sns.histplot(data, bins=10,kde= True, color = "Orange") #kde is the line
plt.title("Histogram using Seaborn")
plt.show()
```

```python
#Boxplot
import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y = [20,30,40,50,60]

plt.boxplot([x, y])
plt.xticks([1, 2], ["Group 1", "Group 2"])
plt.title("BoxPlot using matplot")
```

```python
#Boxplot
import seaborn as sns
group1 = [1,2,3,4,5]
group2 = [20,30,40,50,60]

data = [group1,group2]
sns.boxplot(data)
plt.xticks([0, 1], ["Group 1", "Group 2"])
plt.title("BoxPlot using Seaborn")
```
